```julia
# Задание №8

function outer(x, y, op)
    m, n = length(x), length(y)
    C = Matrix{typeof(op(x[1], y[1]))}(undef, m, n)
    for i in 1:m
        for j in 1:n
            C[i, j] = op(x[i], y[j])
        end
    end
    return C
end

x = [1, 2, 3]
y = [10, 20]
println("outer(x, y, +) ="); display(outer(x, y, +))
println("outer(x, y, *) ="); display(outer(x, y, *))
println("outer(x, y, ^) ="); display(outer(x, y, ^))

v = 0:4
A1 = outer(v, v, (a, b) -> a + b)
println("A1 ="); display(A1)

A2 = outer(0:4, 0:4, (a, b) -> a * b^2)
println("\nA2 ="); display(A2)

A3 = outer(0:4, 0:4, (a, b) -> mod(a + b, 5))
println("\nA3 ="); display(A3)

A4 = outer(0:9, 0:9, (a, b) -> mod(a + b, 10))
println("\nA4 ="); display(A4)

A5 = outer(1:9, 1:9, (a, b) -> mod(a - b, 9))
println("\nA5 ="); display(A5)
```

```julia
# Задание №10

using Random
Random.seed!(42)  

M = rand(1:10, 6, 6)
println("M ="); display(M)

N = 4      
M_val = 7  
K = 75    

count_per_row = [count(>(N), M[i, :]) for i in 1:6]
println("\nЧисло элементов > $N в каждой строке:")
for i in 1:6
    println("  строка $i: $(count_per_row[i])")
end

rows_with_two = [i for i in 1:6 if count(==(M_val), M[i, :]) == 2]
println("\nСтроки, где $M_val встречается ровно 2 раза: ", rows_with_two)

col_sums = vec(sum(M, dims=1))
println("\nСуммы столбцов: ", col_sums)

pairs = Tuple{Int,Int}[]
for j1 in 1:6
    for j2 in (j1+1):6
        if col_sums[j1] + col_sums[j2] > K
            push!(pairs, (j1, j2))
        end
    end
end

println("Пары столбцов с суммой > $K: ", pairs)
```
