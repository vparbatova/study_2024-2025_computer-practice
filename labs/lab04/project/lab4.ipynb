```julia
# Задаём квадратную матрицу 3x3 со случайными значениями:
A = rand(3, 3)

# Задаём единичный вектор:
x = fill(1.0, 3)

# Задаём вектор b:
b = A*x

# Решение исходного уравнения получаем с помощью функции \
# (убеждаемся, что x - единичный вектор):
A\b

# LU- факторизация:
Alu = lu(A)

# Матрица перестановок:
Alu.P

# Вектор перестановок:
Alu.p

# Матрица L:
Alu.L

# Матрица U:
Alu.U

# Решение СЛАУ через матрицу A:
A\b

# Решение СЛАУ через объект факторизации:
Alu\b

# Определитель матрицы A:
det(A)

# Определитель матрицы A через объект факторизации:
det(Alu)

# QR-факторизация:
Aqr = qr(A)

# Матрица Q:
Aqr.Q

# Матрица R:
Aqr.R

# Проверка, что матрица Q - ортогональная:
Aqr.Q' * Aqr.Q
```

```julia
# Задание №2
using LinearAlgebra

# a)
#  x + y = 2,
#  x - y = 3.
A = [1  1;
     1 -1]
b = [2, 3]
x = A \ b
println("1a: x = ", x)   # [2.5, -0.5]

# b)
#  x + y = 2,
#  2x + 2y = 4.   -> бесконечно много решений
A = [1 1;
     2 2]
b = [2, 4]
println("1b: rank(A) = ", rank(A), ", rank([A b]) = ", rank([A b]))
# Ранг A = ранг расширенной = 1 < 2 -> бесконечно много решений
# Частное решение:
x = pinv(A) * b
println("1b: частное решение x = ", x)

# c)
#  x + y = 2,
#  2x + 2y = 5.   -> нет решений
A = [1 1;
     2 2]
b = [2, 5]
println("1c: rank(A) = ", rank(A), ", rank([A b]) = ", rank([A b]))
# rank(A)=1, rank([A b])=2 -> система несовместна

# d)
#  x + y = 1,
#  2x + 2y = 2,
#  3x + 3y = 3.   -> бесконечно много решений
A = [1 1;
     2 2;
     3 3]
b = [1, 2, 3]
println("1d: rank(A) = ", rank(A), ", rank([A b]) = ", rank([A b]))
x = pinv(A) * b
println("1d: частное решение x = ", x)

# e)
#  x + y = 2,
#  2x + y = 1,
#  x - y = 3.   -> переопределённая, проверяем совместность
A = [1  1;
     2  1;
     1 -1]
b = [2, 1, 3]
println("1e: rank(A) = ", rank(A), ", rank([A b]) = ", rank([A b]))
# Если rank(A) == rank([A b]) == 2 -> единственное решение
x = A \ b
println("1e: x = ", x)

# f)
#  x + y = 2,
#  2x + y = 1,
#  3x + 2y = 3.
A = [1 1;
     2 1;
     3 2]
b = [2, 1, 3]
println("1f: rank(A) = ", rank(A), ", rank([A b]) = ", rank([A b]))
x = A \ b
println("1f: x = ", x)

# 2. СЛАУ с тремя неизвестными

# a)
#  x + y + z = 2,
#  x - y - 2z = 3.
A = [1  1  1;
     1 -1 -2]
b = [2, 3]
println("2a: rank(A) = ", rank(A), ", rank([A b]) = ", rank([A b]))
# 2 уравнения, 3 неизвестных -> бесконечно много решений
x = pinv(A) * b
println("2a: частное решение x = ", x)

# b)
#  x + y + z = 2,
#  2x + 2y - 3z = 4,
#  3x + y + z = 1.
A = [1 1  1;
     2 2 -3;
     3 1  1]
b = [2, 4, 1]
println("2b: det(A) = ", det(A))
x = A \ b
println("2b: x = ", x)

# c)
#  x + y + z = 1,
#  x + y + 2z = 0,
#  2x + 2y + 3z = 1.
A = [1 1 1;
     1 1 2;
     2 2 3]
b = [1, 0, 1]
println("2c: rank(A) = ", rank(A), ", rank([A b]) = ", rank([A b]))
x = pinv(A) * b
println("2c: частное решение x = ", x)

# d)
#  x + y + z = 1,
#  x + y + 2z = 0,
#  2x + 2y + 3z = 0.
A = [1 1 1;
     1 1 2;
     2 2 3]
b = [1, 0, 0]
println("2d: rank(A) = ", rank(A), ", rank([A b]) = ", rank([A b]))
# rank(A)=2, rank([A b])=3 -> система несовместна
```

```julia
# Задание №3

using LinearAlgebra

# a)
A = [1 -2;
    -2  1]

# Собственные значения и собственные векторы
E = eigen(A)
println("1a: собственные значения = ", E.values)   # [-1, 3]
println("1a: собственные векторы =\n", E.vectors)

# Диагонализация: A = V * Λ * V^(-1)
V = E.vectors
Λ = Diagonal(E.values)
A_diag = V * Λ * inv(V)
println("1a: проверка V*Λ*V^(-1) =\n", A_diag)


# b)
#  Матрица из задания b) (восстановлена по смыслу)
B = [3 1;
     1 3]

E = eigen(B)
println("1b: собственные значения = ", E.values)   # [2, 4]
println("1b: собственные векторы =\n", E.vectors)

V = E.vectors
Λ = Diagonal(E.values)
println("1b: V*Λ*V^(-1) =\n", V * Λ * inv(V))


# c)
#  Матрица из задания c)
C = [1 2;
     2 1]

E = eigen(C)
println("1c: собственные значения = ", E.values)   # [-1, 3]
println("1c: собственные векторы =\n", E.vectors)

V = E.vectors
Λ = Diagonal(E.values)
println("1c: V*Λ*V^(-1) =\n", V * Λ * inv(V))
```

```julia
# Задание №4

using LinearAlgebra

# Функция: проверяем продуктивность, решая (E - A) x = y
# для базисных неотрицательных y (столбцы единичной матрицы)
function is_productive_def(A)
    n = size(A, 1)
    E = Matrix{Float64}(I, n, n)
    M = E - A
    # Проверяем для каждого базисного вектора y = e_i
    for i in 1:n
        y = zeros(n)
        y[i] = 1.0
        x = M \ y
        if any(x .< -1e-12)   # допуск на погрешность
            return false
        end
    end
    return true
end

# a)
A1 = [1 2;
      3 4]
println("1a: A =\n", A1, "\nпродуктивна? ", is_productive_def(A1))

# b)
A2 = (1/2) * [1 2;
              3 4]
println("\n1b: A =\n", A2, "\nпродуктивна? ", is_productive_def(A2))

# c)
A3 = (1/10) * [1 2;
               3 4]
println("\n1c: A =\n", A3, "\nпродуктивна? ", is_productive_def(A3))


# ============================================================
# 2. Критерий продуктивности: матрица A продуктивна тогда и
#    только тогда, когда все элементы (E - A)^(-1)
#    неотрицательные.
# ============================================================

println("\n=== 2. Критерий (E - A)^(-1) >= 0 ===")

function is_productive_inv(A)
    n = size(A, 1)
    E = Matrix{Float64}(I, n, n)
    M = E - A
    # Проверяем обратимость
    if abs(det(M)) < 1e-12
        return false, nothing
    end
    Minv = inv(M)
    ok = all(Minv .>= -1e-12)
    return ok, Minv
end

# a)
A1 = [1 2;
      3 1]
ok, Minv = is_productive_inv(A1)
println("2a: (E-A)^(-1) =\n", Minv, "\nпродуктивна? ", ok)

# b)
A2 = (1/2) * [1 2;
              3 1]
ok, Minv = is_productive_inv(A2)
println("\n2b: (E-A)^(-1) =\n", Minv, "\nпродуктивна? ", ok)

# c)
A3 = (1/10) * [1 2;
               3 1]
ok, Minv = is_productive_inv(A3)
println("\n2c: (E-A)^(-1) =\n", Minv, "\nпродуктивна? ", ok)


# ============================================================
# 3. Спектральный критерий: матрица A продуктивна тогда и
#    только тогда, когда все её собственные значения
#    по модулю меньше 1.
# ============================================================

println("\n=== 3. Спектральный критерий: max|λ| < 1 ===")

function is_productive_spec(A)
    λ = eigvals(A)
    max_abs = maximum(abs.(λ))
    return max_abs < 1, λ, max_abs
end

# a)
A1 = [1 2;
      3 1]
ok, λ, m = is_productive_spec(A1)
println("3a: λ = ", λ, ", max|λ| = ", m, " -> продуктивна? ", ok)

# b)
A2 = (1/2) * [1 2;
              3 1]
ok, λ, m = is_productive_spec(A2)
println("\n3b: λ = ", λ, ", max|λ| = ", m, " -> продуктивна? ", ok)

# c)
A3 = (1/10) * [1 2;
               3 1]
ok, λ, m = is_productive_spec(A3)
println("\n3c: λ = ", λ, ", max|λ| = ", m, " -> продуктивна? ", ok)

# d) матрица 3x3
A4 = [0.1 0.2 0.3;
      0.0 0.1 0.2;
      0.0 0.1 0.3]
ok, λ, m = is_productive_spec(A4)
println("\n3d: λ = ", λ, ", max|λ| = ", m, " -> продуктивна? ", ok)
```
